````python
"""
Optimized Floor Plan Generator - Compact & Accessible House Layouts
Deep RL model trained specifically for creating compact, highly accessible residential floor plans

Key Focus:
1. Maximum compactness (minimal wasted space)
2. High accessibility (all rooms easily reachable)
3. Efficient circulation (short travel distances)
4. Logical room adjacencies
5. Residential best practices
"""

import math
import random
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import json
import os
from datetime import datetime
from collections import deque

# Check for PyTorch
try:
    import torch
    import torch.nn as nn
    import torch.optim as optim
    PYTORCH_AVAILABLE = True
    print("✓ PyTorch detected - Using Deep Q-Network (DQN)")
except ImportError:
    PYTORCH_AVAILABLE = False
    print("⚠ PyTorch not found - Using rule-based approach")
    print("  Install with: pip install torch --break-system-packages")


class CompactAccessibleRewardShaper:
    """
    SPECIALIZED reward function for COMPACT & ACCESSIBLE house layouts.
    Heavily weighted towards residential efficiency and accessibility.
    """

    @staticmethod
    def calculate_reward(layout_state, action_taken, constraints):
        """
        Calculate reward optimized for COMPACT + ACCESSIBLE layouts.
        This is the core "DNA" that teaches the AI what makes a good house.
        """
        reward = 0.0
        placed_rooms = layout_state['placed_rooms']

        if not placed_rooms:
            return reward

        plot_width = layout_state['plot_width']
        plot_height = layout_state['plot_height']
        plot_area = plot_width * plot_height

        # ═══════════════════════════════════════════════════════════
        # 1. COMPACTNESS (Maximum weight - this is our PRIMARY goal)
        # ═══════════════════════════════════════════════════════════

        # A. Space Utilization (use every square foot)
        total_room_area = sum(r['width'] * r['height'] for r in placed_rooms)
        utilization = total_room_area / plot_area if plot_area > 0 else 0

        if utilization > 0.85:  # Excellent utilization
            reward += 40
        elif utilization > 0.75:  # Good utilization
            reward += 30
        elif utilization > 0.65:  # Acceptable
            reward += 20
        else:  # Poor - penalize
            reward += utilization * 15

        # B. Bounding Box Compactness (tight clustering)
        if len(placed_rooms) >= 2:
            min_x = min(r['x'] for r in placed_rooms)
            min_y = min(r['y'] for r in placed_rooms)
            max_x = max(r['x'] + r['width'] for r in placed_rooms)
            max_y = max(r['y'] + r['height'] for r in placed_rooms)

            bounding_area = (max_x - min_x) * (max_y - min_y)
            if bounding_area > 0:
                compactness_ratio = total_room_area / bounding_area
                if compactness_ratio > 0.9:  # Very tight
                    reward += 35
                elif compactness_ratio > 0.8:  # Tight
                    reward += 25
                else:
                    reward += compactness_ratio * 20

        # C. Perimeter Efficiency (minimize building perimeter)
        perimeter_score = CompactAccessibleRewardShaper._evaluate_perimeter_efficiency(placed_rooms)
        reward += perimeter_score

        # ═══════════════════════════════════════════════════════════
        # 2. ACCESSIBILITY (Second priority - easy movement)
        # ═══════════════════════════════════════════════════════════

        # A. Room Connectivity (all rooms should be reachable)
        connectivity_score = CompactAccessibleRewardShaper._evaluate_connectivity(placed_rooms)
        reward += connectivity_score * 2  # Double weight

        # B. Central Access Point (hub-like layout)
        if CompactAccessibleRewardShaper._has_central_hub(placed_rooms):
            reward += 25

        # C. Short Travel Distances
        avg_distance = CompactAccessibleRewardShaper._average_room_distance(placed_rooms)
        max_possible_distance = math.sqrt(plot_width**2 + plot_height**2)
        if avg_distance < max_possible_distance * 0.3:  # Very close
            reward += 20
        elif avg_distance < max_possible_distance * 0.5:  # Reasonably close
            reward += 10

        # D. No Isolated Rooms (every room adjacent to at least one other)
        if CompactAccessibleRewardShaper._no_isolated_rooms(placed_rooms):
            reward += 15

        # ═══════════════════════════════════════════════════════════
        # 3. RESIDENTIAL LOGIC (house-specific rules)
        # ═══════════════════════════════════════════════════════════

        # A. Essential Adjacencies for Houses
        adjacency_bonus = CompactAccessibleRewardShaper._evaluate_house_adjacencies(placed_rooms)
        reward += adjacency_bonus

        # B. Functional Zones (private vs public separation)
        zoning_score = CompactAccessibleRewardShaper._evaluate_functional_zoning(placed_rooms)
        reward += zoning_score

        # C. Traffic Flow (efficient circulation)
        flow_score = CompactAccessibleRewardShaper._evaluate_traffic_flow(placed_rooms)
        reward += flow_score

        # ═══════════════════════════════════════════════════════════
        # 4. CRITICAL PENALTIES (violations)
        # ═══════════════════════════════════════════════════════════

        # A. Overlapping rooms (absolute no-go)
        if CompactAccessibleRewardShaper._has_overlaps(placed_rooms):
            reward -= 100  # Massive penalty

        # B. Scattered layout (gaps between rooms)
        gap_penalty = CompactAccessibleRewardShaper._calculate_gap_penalty(placed_rooms, layout_state)
        reward -= gap_penalty

        # C. Inefficient shapes (too elongated)
        last_room = placed_rooms[-1]
        aspect_ratio = max(last_room['width'], last_room['height']) / max(min(last_room['width'], last_room['height']), 1)
        if aspect_ratio > 3.0:  # Too thin/elongated
            reward -= 15
        elif 1.0 <= aspect_ratio <= 2.0:  # Good proportion
            reward += 8

        # D. Minimum size requirements
        if 'min_size' in last_room:
            actual_area = last_room['width'] * last_room['height']
            if actual_area < last_room['min_size']:
                reward -= 20
            else:
                reward += 5

        # ═══════════════════════════════════════════════════════════
        # 5. CONSTRAINT SATISFACTION
        # ═══════════════════════════════════════════════════════════
        constraint_score = CompactAccessibleRewardShaper._evaluate_constraints(placed_rooms, constraints)
        reward += constraint_score * 1.5  # Boost constraint satisfaction

        return reward

    @staticmethod
    def _evaluate_perimeter_efficiency(placed_rooms):
        """
        Reward layouts with minimal external perimeter (more compact).
        Less perimeter = more efficient building envelope.
        """
        if len(placed_rooms) < 2:
            return 0.0

        # Calculate shared walls (good - reduces perimeter)
        shared_wall_length = 0
        for i, r1 in enumerate(placed_rooms):
            for r2 in placed_rooms[i+1:]:
                shared = CompactAccessibleRewardShaper._calculate_shared_wall(r1, r2)
                shared_wall_length += shared

        # More shared walls = higher score
        return min(shared_wall_length * 0.5, 20)

    @staticmethod
    def _calculate_shared_wall(room1, room2):
        """Calculate length of shared wall between two rooms"""
        x1, y1, w1, h1 = room1['x'], room1['y'], room1['width'], room1['height']
        x2, y2, w2, h2 = room2['x'], room2['y'], room2['width'], room2['height']

        # Check vertical shared wall
        if abs((x1 + w1) - x2) < 1.0 or abs((x2 + w2) - x1) < 1.0:
            overlap_start = max(y1, y2)
            overlap_end = min(y1 + h1, y2 + h2)
            if overlap_end > overlap_start:
                return overlap_end - overlap_start

        # Check horizontal shared wall
        if abs((y1 + h1) - y2) < 1.0 or abs((y2 + h2) - y1) < 1.0:
            overlap_start = max(x1, x2)
            overlap_end = min(x1 + w1, x2 + w2)
            if overlap_end > overlap_start:
                return overlap_end - overlap_start

        return 0.0

    @staticmethod
    def _evaluate_connectivity(placed_rooms):
        """Reward well-connected layouts (high adjacency)"""
        if len(placed_rooms) < 2:
            return 0.0

        adjacent_pairs = 0
        for i, r1 in enumerate(placed_rooms):
            for r2 in placed_rooms[i+1:]:
                if CompactAccessibleRewardShaper._are_adjacent(r1, r2):
                    adjacent_pairs += 1

        # High connectivity bonus
        max_possible_pairs = len(placed_rooms) * (len(placed_rooms) - 1) / 2
        connectivity_ratio = adjacent_pairs / max_possible_pairs if max_possible_pairs > 0 else 0

        return connectivity_ratio * 30  # Up to +30 points

    @staticmethod
    def _has_central_hub(placed_rooms):
        """Check if there's a central room connected to many others (good for accessibility)"""
        for room in placed_rooms:
            adjacent_count = sum(1 for other in placed_rooms
                               if other != room and CompactAccessibleRewardShaper._are_adjacent(room, other))
            if adjacent_count >= len(placed_rooms) // 2:  # Connected to half or more
                return True
        return False

    @staticmethod
    def _no_isolated_rooms(placed_rooms):
        """Check that no room is isolated (all have at least one neighbor)"""
        if len(placed_rooms) < 2:
            return True

        for room in placed_rooms:
            has_neighbor = any(CompactAccessibleRewardShaper._are_adjacent(room, other)
                             for other in placed_rooms if other != room)
            if not has_neighbor:
                return False
        return True

    @staticmethod
    def _average_room_distance(placed_rooms):
        """Calculate average center-to-center distance between rooms"""
        if len(placed_rooms) < 2:
            return 0.0

        total_distance = 0
        count = 0

        for i, r1 in enumerate(placed_rooms):
            for r2 in placed_rooms[i+1:]:
                cx1 = r1['x'] + r1['width'] / 2
                cy1 = r1['y'] + r1['height'] / 2
                cx2 = r2['x'] + r2['width'] / 2
                cy2 = r2['y'] + r2['height'] / 2
                distance = math.sqrt((cx2 - cx1)**2 + (cy2 - cy1)**2)
                total_distance += distance
                count += 1

        return total_distance / count if count > 0 else 0.0

    @staticmethod
    def _evaluate_house_adjacencies(placed_rooms):
        """Reward residential-specific logical adjacencies"""
        score = 0.0

        # Define CRITICAL house adjacencies
        critical_adjacencies = [
            ('kitchen', 'dining_room', 20),     # Kitchen near dining
            ('living_room', 'kitchen', 15),      # Living near kitchen
            ('bedroom', 'bathroom', 18),         # Bedroom near bathroom
            ('living_room', 'dining_room', 12),  # Living near dining
        ]

        # Define GOOD adjacencies
        good_adjacencies = [
            ('kitchen', 'bathroom', -10),        # Kitchen NOT near bathroom
            ('bedroom', 'living_room', -5),      # Bedroom separated from living
        ]

        for room1_type, room2_type, bonus in critical_adjacencies + good_adjacencies:
            rooms1 = [r for r in placed_rooms if r['type'] == room1_type]
            rooms2 = [r for r in placed_rooms if r['type'] == room2_type]

            for r1 in rooms1:
                for r2 in rooms2:
                    if CompactAccessibleRewardShaper._are_adjacent(r1, r2):
                        score += bonus

        return score

    @staticmethod
    def _evaluate_functional_zoning(placed_rooms):
        """Reward separation of private (bedrooms) and public (living) spaces"""
        private_rooms = [r for r in placed_rooms if r['type'] in ['bedroom', 'bathroom']]
        public_rooms = [r for r in placed_rooms if r['type'] in ['living_room', 'kitchen', 'dining_room']]

        if not private_rooms or not public_rooms:
            return 0.0

        score = 0.0

        # Private rooms should cluster together
        if len(private_rooms) >= 2:
            for i, r1 in enumerate(private_rooms):
                for r2 in private_rooms[i+1:]:
                    if CompactAccessibleRewardShaper._are_adjacent(r1, r2):
                        score += 8  # Bonus for private clustering

        # Public rooms should cluster together
        if len(public_rooms) >= 2:
            for i, r1 in enumerate(public_rooms):
                for r2 in public_rooms[i+1:]:
                    if CompactAccessibleRewardShaper._are_adjacent(r1, r2):
                        score += 8  # Bonus for public clustering

        return score

    @staticmethod
    def _evaluate_traffic_flow(placed_rooms):
        """Reward efficient circulation patterns"""
        if len(placed_rooms) < 3:
            return 0.0

        # Check for linear vs branching flow
        # Linear (hallway-like): lower score
        # Branching (hub): higher score

        adjacency_counts = []
        for room in placed_rooms:
            count = sum(1 for other in placed_rooms
                       if other != room and CompactAccessibleRewardShaper._are_adjacent(room, other))
            adjacency_counts.append(count)

        # Prefer layouts where rooms have 2-3 connections (good flow)
        good_flow_rooms = sum(1 for count in adjacency_counts if 2 <= count <= 3)

        return good_flow_rooms * 5

    @staticmethod
    def _calculate_gap_penalty(placed_rooms, layout_state):
        """Penalize gaps/wasted space between rooms"""
        if len(placed_rooms) < 2:
            return 0.0

        # Calculate total "empty" space in bounding box
        min_x = min(r['x'] for r in placed_rooms)
        min_y = min(r['y'] for r in placed_rooms)
        max_x = max(r['x'] + r['width'] for r in placed_rooms)
        max_y = max(r['y'] + r['height'] for r in placed_rooms)

        bounding_area = (max_x - min_x) * (max_y - min_y)
        total_room_area = sum(r['width'] * r['height'] for r in placed_rooms)

        gap_area = bounding_area - total_room_area
        gap_ratio = gap_area / bounding_area if bounding_area > 0 else 0

        # Penalize based on gap ratio
        if gap_ratio > 0.3:  # More than 30% gaps
            return gap_ratio * 30
        elif gap_ratio > 0.15:  # 15-30% gaps
            return gap_ratio * 15

        return 0.0

    @staticmethod
    def _are_adjacent(room1, room2, tolerance=2.0):
        """Check if two rooms share a wall"""
        x1, y1, w1, h1 = room1['x'], room1['y'], room1['width'], room1['height']
        x2, y2, w2, h2 = room2['x'], room2['y'], room2['width'], room2['height']

        # X-axis adjacency
        if (abs((x1 + w1) - x2) < tolerance or abs((x2 + w2) - x1) < tolerance) and \
           max(y1, y2) < min(y1 + h1, y2 + h2):
            return True

        # Y-axis adjacency
        if (abs((y1 + h1) - y2) < tolerance or abs((y2 + h2) - y1) < tolerance) and \
           max(x1, x2) < min(x1 + w1, x2 + w2):
            return True

        return False

    @staticmethod
    def _has_overlaps(placed_rooms):
        """Check for any overlapping rooms"""
        for i, room1 in enumerate(placed_rooms):
            for room2 in placed_rooms[i+1:]:
                if CompactAccessibleRewardShaper._are_overlapping(
                    (room1['x'], room1['y'], room1['width'], room1['height']),
                    (room2['x'], room2['y'], room2['width'], room2['height'])):
                    return True
        return False

    @staticmethod
    def _are_overlapping(rect1, rect2):
        """Check if rectangles overlap"""
        x1, y1, w1, h1 = rect1
        x2, y2, w2, h2 = rect2
        return not (x1 >= x2 + w2 or x2 >= x1 + w1 or y1 >= y2 + h2 or y2 >= y1 + h1)

    @staticmethod
    def _evaluate_constraints(placed_rooms, constraints):
        """Reward user constraint satisfaction"""
        score = 0.0

        for constraint in constraints:
            if constraint['type'] == 'adjacency':
                room1_type = constraint['details']['room1']
                room2_type = constraint['details']['room2']

                rooms1 = [r for r in placed_rooms if r['type'] == room1_type]
                rooms2 = [r for r in placed_rooms if r['type'] == room2_type]

                for r1 in rooms1:
                    for r2 in rooms2:
                        if CompactAccessibleRewardShaper._are_adjacent(r1, r2):
                            score += 25  # Big bonus

        return score


class LayoutQNetwork(nn.Module):
    """Deep Q-Network for learning optimal room placements"""
    def __init__(self, state_size, action_size, hidden_size=256):
        super(LayoutQNetwork, self).__init__()

        # Network without BatchNorm (causes issues with batch_size=1)
        self.network = nn.Sequential(
            nn.Linear(state_size, hidden_size),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(hidden_size, hidden_size),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(hidden_size, hidden_size // 2),
            nn.ReLU(),

            nn.Linear(hidden_size // 2, action_size)
        )

    def forward(self, x):
        return self.network(x)


class DQNAgent:
    """Deep Q-Learning Agent optimized for compact layouts"""
    def __init__(self, state_size, action_size, learning_rate=0.0005):
        self.state_size = state_size
        self.action_size = action_size
        self.memory = deque(maxlen=5000)  # Larger memory

        # Optimized hyperparameters for compact layouts
        self.gamma = 0.97  # Higher discount (long-term focus)
        self.epsilon = 1.0
        self.epsilon_min = 0.05  # Keep some exploration
        self.epsilon_decay = 0.992
        self.learning_rate = learning_rate
        self.batch_size = 64  # Larger batches

        if PYTORCH_AVAILABLE:
            self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
            self.model = LayoutQNetwork(state_size, action_size, hidden_size=256).to(self.device)
            self.target_model = LayoutQNetwork(state_size, action_size, hidden_size=256).to(self.device)
            self.optimizer = optim.Adam(self.model.parameters(), lr=learning_rate, weight_decay=1e-5)
            self.criterion = nn.MSELoss()
            self.update_target_model()

    def update_target_model(self):
        if PYTORCH_AVAILABLE:
            self.target_model.load_state_dict(self.model.state_dict())

    def remember(self, state, action, reward, next_state, done):
        self.memory.append((state, action, reward, next_state, done))

    def act(self, state):
        if np.random.random() <= self.epsilon:
            return random.randrange(self.action_size)

        if PYTORCH_AVAILABLE:
            state_tensor = torch.FloatTensor(state).unsqueeze(0).to(self.device)
            with torch.no_grad():
                q_values = self.model(state_tensor)
            return q_values.argmax().item()
        else:
            return random.randrange(self.action_size)

    def replay(self):
        if len(self.memory) < self.batch_size:
            return 0.0

        if not PYTORCH_AVAILABLE:
            return 0.0

        batch = random.sample(self.memory, self.batch_size)

        states = torch.FloatTensor([e[0] for e in batch]).to(self.device)
        actions = torch.LongTensor([e[1] for e in batch]).to(self.device)
        rewards = torch.FloatTensor([e[2] for e in batch]).to(self.device)
        next_states = torch.FloatTensor([e[3] for e in batch]).to(self.device)
        dones = torch.FloatTensor([e[4] for e in batch]).to(self.device)

        current_q_values = self.model(states).gather(1, actions.unsqueeze(1))

        with torch.no_grad():
            next_q_values = self.target_model(next_states).max(1)[0]
            target_q_values = rewards + (1 - dones) * self.gamma * next_q_values

        loss = self.criterion(current_q_values.squeeze(), target_q_values)

        self.optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(self.model.parameters(), 1.0)
        self.optimizer.step()

        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay

        return loss.item()

    def save(self, filepath):
        if PYTORCH_AVAILABLE:
            torch.save({
                'model_state_dict': self.model.state_dict(),
                'target_model_state_dict': self.target_model.state_dict(),
                'optimizer_state_dict': self.optimizer.state_dict(),
                'epsilon': self.epsilon
            }, filepath)

    def load(self, filepath):
        if PYTORCH_AVAILABLE and os.path.exists(filepath):
            checkpoint = torch.load(filepath)
            self.model.load_state_dict(checkpoint['model_state_dict'])
            self.target_model.load_state_dict(checkpoint['target_model_state_dict'])
            self.optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
            self.epsilon = checkpoint['epsilon']
            print(f"✓ Loaded pre-trained model (ε={self.epsilon:.3f})")


class CompactHouseGenerator:
    """
    Specialized generator for COMPACT, ACCESSIBLE house layouts.
    Pre-configured for residential efficiency.
    """

    # Standard room size ratios (as % of total area)
    ROOM_SIZE_STANDARDS = {
        'living_room': {'min_ratio': 0.15, 'target_ratio': 0.20},
        'kitchen': {'min_ratio': 0.08, 'target_ratio': 0.12},
        'dining_room': {'min_ratio': 0.08, 'target_ratio': 0.11},
        'bedroom': {'min_ratio': 0.09, 'target_ratio': 0.13},
        'master_bedroom': {'min_ratio': 0.12, 'target_ratio': 0.16},
        'bathroom': {'min_ratio': 0.03, 'target_ratio': 0.05},
        'master_bathroom': {'min_ratio': 0.04, 'target_ratio': 0.06},
        'office': {'min_ratio': 0.07, 'target_ratio': 0.10},
        'laundry': {'min_ratio': 0.02, 'target_ratio': 0.04},
        'closet': {'min_ratio': 0.02, 'target_ratio': 0.03},
        'hallway': {'min_ratio': 0.03, 'target_ratio': 0.05},
    }

    def __init__(self, width, height, use_ml=True):
        self.width = int(width)
        self.height = int(height)
        self.room_requirements = []
        self.constraints = []
        self.use_ml = use_ml and PYTORCH_AVAILABLE

        if self.use_ml:
            state_size = 25
            action_size = 144  # 12x12 grid for finer placement
            self.agent = DQNAgent(state_size, action_size, learning_rate=0.0005)

            model_path = 'compact_house_model.pth'
            if os.path.exists(model_path):
                self.agent.load(model_path)

        print(f"CompactHouseGenerator: {width}x{height}")
        print("✓ Optimized for COMPACT & ACCESSIBLE residential layouts")

    def auto_add_room(self, room_type, count=1):
        """
        Automatically add room with standard sizes based on plot area.

        Args:
            room_type: Type of room (e.g., 'bedroom', 'kitchen')
            count: Number of this room type to add
        """
        plot_area = self.width * self.height

        # Get standard ratios or use defaults
        if room_type in self.ROOM_SIZE_STANDARDS:
            min_ratio = self.ROOM_SIZE_STANDARDS[room_type]['min_ratio']
            target_ratio = self.ROOM_SIZE_STANDARDS[room_type]['target_ratio']
        else:
            # Default for unknown room types
            min_ratio = 0.07
            target_ratio = 0.10

        # Calculate actual sizes
        min_size = plot_area * min_ratio
        target_area = plot_area * target_ratio

        # Add room(s)
        for i in range(count):
            self.room_requirements.append({
                'room_type': room_type,
                'min_size': min_size,
                'target_area': target_area
            })
            print(f"✓ Added {room_type}: min={min_size:.0f} sq ft, target={target_area:.0f} sq ft")

    def auto_add_constraints(self):
        """
        Automatically add logical constraints based on room types present.
        This ensures residential best practices.
        """
        room_types_present = set(req['room_type'] for req in self.room_requirements)

        # Define smart adjacencies
        auto_constraints = []

        # Kitchen + Dining Room
        if 'kitchen' in room_types_present and 'dining_room' in room_types_present:
            auto_constraints.append(('adjacency', {'room1': 'kitchen', 'room2': 'dining_room'}))

        # Kitchen + Living Room
        if 'kitchen' in room_types_present and 'living_room' in room_types_present:
            auto_constraints.append(('adjacency', {'room1': 'kitchen', 'room2': 'living_room'}))

        # Living Room + Dining Room
        if 'living_room' in room_types_present and 'dining_room' in room_types_present:
            auto_constraints.append(('adjacency', {'room1': 'living_room', 'room2': 'dining_room'}))

        # Bedroom + Bathroom
        if 'bedroom' in room_types_present and 'bathroom' in room_types_present:
            auto_constraints.append(('adjacency', {'room1': 'bedroom', 'room2': 'bathroom'}))

        # Master Bedroom + Master Bathroom
        if 'master_bedroom' in room_types_present and 'master_bathroom' in room_types_present:
            auto_constraints.append(('adjacency', {'room1': 'master_bedroom', 'room2': 'master_bathroom'}))

        # Add constraints
        for ctype, details in auto_constraints:
            self.add_constraint(ctype, details)
            print(f"✓ Auto-constraint: {details['room1']} ↔ {details['room2']}")

    def add_room_requirement(self, room_type, min_size, target_area):
        self.room_requirements.append({
            'room_type': room_type,
            'min_size': min_size,
            'target_area': target_area
        })

    def add_constraint(self, constraint_type, details):
        self.constraints.append({
            'type': constraint_type,
            'details': details
        })

    def _extract_state_features(self, placed_rooms, current_room_req):
        """Extract comprehensive state features"""
        features = []

        total_room_area = sum(r['width'] * r['height'] for r in placed_rooms)
        plot_area = self.width * self.height

        # Utilization
        features.append(total_room_area / plot_area if plot_area > 0 else 0)

        # Room count
        features.append(len(placed_rooms) / 10.0)

        # Current room specs
        features.append(current_room_req['target_area'] / plot_area)
        features.append(current_room_req['min_size'] / plot_area)

        # Bounding box
        if placed_rooms:
            min_x = min(r['x'] for r in placed_rooms)
            min_y = min(r['y'] for r in placed_rooms)
            max_x = max(r['x'] + r['width'] for r in placed_rooms)
            max_y = max(r['y'] + r['height'] for r in placed_rooms)
            features.append(max_x / self.width)
            features.append(max_y / self.height)

            # Compactness ratio
            bounding_area = (max_x - min_x) * (max_y - min_y)
            features.append(total_room_area / bounding_area if bounding_area > 0 else 0)
        else:
            features.extend([0.0, 0.0, 0.0])

        # Room type encoding
        room_types = ['living_room', 'kitchen', 'bedroom', 'bathroom', 'dining_room', 'office']
        for rt in room_types:
            features.append(1.0 if current_room_req['room_type'] == rt else 0.0)

        # Connectivity
        if len(placed_rooms) >= 2:
            adjacent_count = sum(1 for i, r1 in enumerate(placed_rooms)
                               for r2 in placed_rooms[i+1:]
                               if CompactAccessibleRewardShaper._are_adjacent(r1, r2))
            features.append(adjacent_count / len(placed_rooms))
        else:
            features.append(0.0)

        # Average distance
        if len(placed_rooms) >= 2:
            avg_dist = CompactAccessibleRewardShaper._average_room_distance(placed_rooms)
            max_dist = math.sqrt(self.width**2 + self.height**2)
            features.append(avg_dist / max_dist if max_dist > 0 else 0)
        else:
            features.append(0.0)

        # Constraint indicators
        features.append(len(self.constraints) / 5.0)

        # Private/public room counts
        private_count = sum(1 for r in placed_rooms if r['type'] in ['bedroom', 'bathroom'])
        public_count = sum(1 for r in placed_rooms if r['type'] in ['living_room', 'kitchen', 'dining_room'])
        features.append(private_count / 5.0)
        features.append(public_count / 5.0)

        # Isolated rooms check
        isolated = sum(1 for r in placed_rooms
                      if not any(CompactAccessibleRewardShaper._are_adjacent(r, other)
                               for other in placed_rooms if other != r))
        features.append(isolated / max(len(placed_rooms), 1))

        while len(features) < 25:
            features.append(0.0)

        return np.array(features[:25], dtype=np.float32)

    def _action_to_position(self, action, room_width, room_height):
        grid_size = 12
        grid_x = action % grid_size
        grid_y = action // grid_size

        x = int((grid_x / grid_size) * (self.width - room_width))
        y = int((grid_y / grid_size) * (self.height - room_height))

        return max(0, x), max(0, y)

    def _is_valid_placement(self, x, y, width, height, placed_rooms):
        if x < 0 or y < 0 or x + width > self.width or y + height > self.height:
            return False

        new_room_bbox = (x, y, width, height)
        for room in placed_rooms:
            existing_bbox = (room['x'], room['y'], room['width'], room['height'])
            if CompactAccessibleRewardShaper._are_overlapping(new_room_bbox, existing_bbox):
                return False

        return True

    def generate_compact_layout(self, training_episodes=150):
        """Generate compact, accessible layout with RL"""
        print(f"\n{'='*60}")
        print(f"  Training for COMPACT & ACCESSIBLE layouts")
        print(f"  Episodes: {training_episodes}")
        print(f"{'='*60}\n")

        best_layout = None
        best_reward = -float('inf')
        episode_rewards = []

        for episode in range(training_episodes):
            placed_rooms = []
            episode_reward = 0
            room_counter = 0

            # Sort by area (largest first) for better packing
            sorted_reqs = sorted(self.room_requirements, key=lambda r: r['target_area'], reverse=True)

            for room_req in sorted_reqs:
                state = self._extract_state_features(placed_rooms, room_req)

                target_area = room_req['target_area']
                room_width = int(math.sqrt(target_area))
                room_height = int(target_area / room_width) if room_width > 0 else int(math.sqrt(target_area))

                while room_width * room_height < room_req['min_size']:
                    if room_width <= room_height:
                        room_width += 1
                    else:
                        room_height += 1

                if self.use_ml:
                    action = self.agent.act(state)
                else:
                    action = random.randint(0, 143)

                x, y = self._action_to_position(action, room_width, room_height)

                attempts = 0
                while not self._is_valid_placement(x, y, room_width, room_height, placed_rooms) and attempts < 30:
                    action = random.randint(0, 143)
                    x, y = self._action_to_position(action, room_width, room_height)
                    attempts += 1

                if attempts >= 30:
                    reward = -50
                    done = True
                else:
                    room = {
                        'id': f"{room_req['room_type']}_{room_counter:03d}",
                        'type': room_req['room_type'],
                        'x': x,
                        'y': y,
                        'width': room_width,
                        'height': room_height,
                        'min_size': room_req['min_size']
                    }
                    placed_rooms.append(room)
                    room_counter += 1

                    layout_state = {
                        'placed_rooms': placed_rooms,
                        'plot_width': self.width,
                        'plot_height': self.height
                    }
                    reward = CompactAccessibleRewardShaper.calculate_reward(
                        layout_state, action, self.constraints
                    )
                    done = False

                next_state = self._extract_state_features(placed_rooms, room_req)

                if self.use_ml:
                    self.agent.remember(state, action, reward, next_state, done)
                    self.agent.replay()

                episode_reward += reward

                if done:
                    break

            episode_rewards.append(episode_reward)

            if episode_reward > best_reward and len(placed_rooms) == len(self.room_requirements):
                best_reward = episode_reward
                best_layout = placed_rooms

                if episode % 10 == 0:
                    print(f"  Ep {episode}: ★ NEW BEST = {best_reward:.1f} | ε={self.agent.epsilon:.3f}")
            elif episode % 20 == 0:
                recent_avg = np.mean(episode_rewards[-20:])
                print(f"  Ep {episode}: Avg(20) = {recent_avg:.1f}")

            if self.use_ml and episode % 10 == 0:
                self.agent.update_target_model()

        if self.use_ml:
            self.agent.save('compact_house_model.pth')
            print(f"\n✓ Model saved")

        doors = self._generate_doors(best_layout) if best_layout else []

        layout_data = {
            'floor_id': 'compact_house_1',
            'plot_dimensions': {'width': self.width, 'height': self.height},
            'rooms': best_layout if best_layout else [],
            'doors': doors,
            'corridors': [],
            'stairs': [],
            'score': best_reward,
            'training_episodes': training_episodes
        }

        print(f"\n{'='*60}")
        print(f"  ✓ COMPACT HOUSE GENERATION COMPLETE")
        print(f"  Best Reward: {best_reward:.1f}")
        print(f"  Rooms: {len(best_layout) if best_layout else 0}/{len(self.room_requirements)}")
        print(f"{'='*60}\n")

        return layout_data, episode_rewards

    def _generate_doors(self, placed_rooms):
        doors = []
        door_counter = 0

        for i, room1 in enumerate(placed_rooms):
            for room2 in placed_rooms[i+1:]:
                if CompactAccessibleRewardShaper._are_adjacent(room1, room2):
                    x1, y1, w1, h1 = room1['x'], room1['y'], room1['width'], room1['height']
                    x2, y2, w2, h2 = room2['x'], room2['y'], room2['width'], room2['height']

                    door_x = (x1 + w1/2 + x2 + w2/2) / 2
                    door_y = (y1 + h1/2 + y2 + h2/2) / 2

                    doors.append({
                        'id': f'door_{door_counter:03d}',
                        'x': door_x,
                        'y': door_y,
                        'width': 1,
                        'height': 1,
                        'connects': [room1['id'], room2['id']]
                    })
                    door_counter += 1

        return doors


def plot_compact_house(layout, episode_rewards):
    """Specialized visualization for compact houses"""
    fig = plt.figure(figsize=(16, 7))

    # Floor plan
    ax1 = plt.subplot(1, 2, 1)

    plot_width = layout['plot_dimensions']['width']
    plot_height = layout['plot_dimensions']['height']
    ax1.set_xlim(0, plot_width)
    ax1.set_ylim(0, plot_height)
    ax1.set_aspect('equal')

    plot_boundary = patches.Rectangle((0, 0), plot_width, plot_height,
                                      linewidth=3, edgecolor='#2C3E50', facecolor='#ECF0F1')
    ax1.add_patch(plot_boundary)

    colors = {
        'living_room': '#3498DB', 'kitchen': '#E67E22', 'bedroom': '#9B59B6',
        'bathroom': '#1ABC9C', 'dining_room': '#E74C3C', 'office': '#F39C12',
        'default': '#95A5A6'
    }

    for room in layout['rooms']:
        color = colors.get(room['type'], colors['default'])
        rect = patches.Rectangle((room['x'], room['y']), room['width'], room['height'],
                                 linewidth=2.5, edgecolor='white', facecolor=color, alpha=0.8)
        ax1.add_patch(rect)

        ax1.text(room['x'] + room['width']/2, room['y'] + room['height']/2,
                room['type'].replace('_', '\n').title(),
                ha='center', va='center', fontsize=9, weight='bold', color='white')

        ax1.text(room['x'] + room['width']/2, room['y'] + room['height']/2 - 1.8,
                f"{room['width']*room['height']:.0f} sq ft",
                ha='center', va='center', fontsize=7, color='white', style='italic')

    for door in layout['doors']:
        circle = patches.Circle((door['x'], door['y']), 1.0,
                               edgecolor='#34495E', facecolor='#BDC3C7', linewidth=2)
        ax1.add_patch(circle)

    # Calculate metrics
    total_area = sum(r['width'] * r['height'] for r in layout['rooms'])
    utilization = (total_area / (plot_width * plot_height)) * 100

    ax1.set_title(f"Compact House Layout\nUtilization: {utilization:.1f}% | Reward: {layout['score']:.1f}",
                 fontsize=13, weight='bold', pad=15)
    ax1.set_xlabel("Width (feet)", fontsize=10)
    ax1.set_ylabel("Height (feet)", fontsize=10)
    ax1.grid(True, linestyle=':', alpha=0.3, color='gray')
    ax1.invert_yaxis()

    # Learning curve
    ax2 = plt.subplot(1, 2, 2)

    if episode_rewards:
        episodes = range(1, len(episode_rewards) + 1)
        ax2.plot(episodes, episode_rewards, alpha=0.4, color='#3498DB', label='Episode Reward')

        window = 20
        if len(episode_rewards) >= window:
            moving_avg = np.convolve(episode_rewards, np.ones(window)/window, mode='valid')
            ax2.plot(range(window, len(episode_rewards) + 1), moving_avg,
                    color='#E74C3C', linewidth=2.5, label=f'Trend (MA-{window})')

        ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
        ax2.set_xlabel('Training Episode', fontsize=10)
        ax2.set_ylabel('Cumulative Reward', fontsize=10)
        ax2.set_title('AI Learning Progress\n(Improving Layout Quality)', fontsize=13, weight='bold')
        ax2.legend(loc='lower right')
        ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


def get_numeric_input(prompt, type_func=float):
    while True:
        try:
            value = type_func(input(prompt))
            if value <= 0:
                print("Must be positive.")
                continue
            return value
        except ValueError:
            print("Invalid input.")


def get_simple_room_input():
    """Simplified input - just ask for room types and counts"""
    print("\n" + "="*60)
    print("  ROOM CONFIGURATION (Simplified)")
    print("="*60)
    print("\nAvailable room types:")
    print("  • living_room      • kitchen         • dining_room")
    print("  • bedroom          • master_bedroom • bathroom")
    print("  • master_bathroom  • office          • laundry")
    print("\nEnter room types and quantities.")
    print("Room sizes will be automatically calculated based on plot area.\n")

    rooms = []

    while True:
        room_type = input("Room type (or 'done' to finish): ").strip().lower().replace(' ', '_')
        if room_type == 'done':
            break

        if not room_type:
            continue

        # Ask for quantity
        try:
            quantity = int(input(f"  How many {room_type}(s)? [1]: ").strip() or "1")
            if quantity < 1:
                print("  Quantity must be at least 1.")
                continue
        except ValueError:
            print("  Invalid quantity. Using 1.")
            quantity = 1

        # Add the room(s)
        for i in range(quantity):
            rooms.append(room_type)

        print(f"  ✓ Added {quantity} {room_type}(s)")

    return rooms


if __name__ == '__main__':
    print("="*60)
    print("   🏠 COMPACT & ACCESSIBLE House Generator")
    print("   Simplified Input - Automatic Sizing")
    print("="*60)

    if not PYTORCH_AVAILABLE:
        print("\n⚠️  For best results, install PyTorch:")
        print("   pip install torch --break-system-packages\n")

    # Simplified input
    print("\n📐 STEP 1: Plot Dimensions")
    total_area = get_numeric_input("Total plot area (sq ft): ")

    # Auto-calculate dimensions (assume square or 4:3 ratio)
    aspect_ratio = input("Aspect ratio (press Enter for square, or type '4:3', '3:2', etc.): ").strip()

    if not aspect_ratio:
        # Square
        width = height = int(math.sqrt(total_area))
    else:
        try:
            w_ratio, h_ratio = map(float, aspect_ratio.split(':'))
            # width/height = w_ratio/h_ratio and width*height = total_area
            height = math.sqrt(total_area * h_ratio / w_ratio)
            width = total_area / height
        except:
            print("Invalid ratio, using square.")
            width = height = int(math.sqrt(total_area))

    width = int(width)
    height = int(height)

    print(f"\n✓ Plot dimensions: {width} ft × {height} ft = {width*height} sq ft")

    # Create generator
    generator = CompactHouseGenerator(width, height)

    # Get rooms
    print("\n🏠 STEP 2: Room Configuration")
    room_list = get_simple_room_input()

    if not room_list:
        print("\n⚠️  No rooms specified. Exiting.")
        exit()

    # Auto-add rooms with calculated sizes
    print(f"\n📊 Calculating room sizes based on {width*height} sq ft plot...")
    for room_type in room_list:
        generator.auto_add_room(room_type)

    # Auto-add constraints
    print("\n🔗 Adding automatic constraints...")
    generator.auto_add_constraints()

    # Training episodes
    print("\n🤖 STEP 3: AI Training")
    episodes = get_numeric_input("Training episodes (recommended: 150-250): ", int)

    # Generate layout
    layout, rewards = generator.generate_compact_layout(training_episodes=episodes)

    # Display results
    print("\n" + "="*60)
    print("   FINAL LAYOUT METRICS")
    print("="*60)
    print(f"Reward Score: {layout['score']:.1f}")
    print(f"Rooms Placed: {len(layout['rooms'])}/{len(room_list)}")
    print(f"Doors: {len(layout['doors'])}")

    total_room_area = sum(r['width'] * r['height'] for r in layout['rooms'])
    plot_area = width * height
    print(f"Space Utilization: {(total_room_area/plot_area)*100:.1f}%")

    print("\n📋 Room Details:")
    for room in layout['rooms']:
        print(f"  • {room['type']}: {room['width']}×{room['height']} = {room['width']*room['height']:.0f} sq ft")

    print("\n📊 Generating visualization...")
    plot_compact_house(layout, rewards)

    print("\n✓ Compact house generation complete!")
    print("  Model saved - will be faster next time!")
````
